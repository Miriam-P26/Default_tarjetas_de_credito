3.4 Recodificación de categorías poco frecuentes

**EDUCACIÓN**

Las categorías 0, 5 y 6 presenan una frecuencia significativamene menor respecto de las categorías*principales.
Además, las categorís 5 y 6 se encuentran documentadascomo valores desconocidos.
Por ese motivo se propone reagrupar dichs valores dentro de la categoría 4*(Otros).

**ESTADO CIVIL**

La categoría 0 presenta una frecuecia muy reducida dentro del conjunTo de datos.
Dado que la categoría 3 ya representa estados civiles agupados bajo la denominación "Otros*, se incorpora dentro de dicha categoría.

```python
def limpiar(datos):
    datos = datos.copy()
    datos["educacion"] = datos["educacion"].replace({0: 4, 5: 4, 6: 4})
    datos["estado_civil"] = datos["estado_civil"].replace({0: 3})
    return datos

```

```python
from sklearn.preprocessing import FunctionTransformer
```

```python
preprocess_transform = FunctionTransformer(
    func=limpiar,
    feature_names_out="one-to-one"
)
```

Para preparar correctamente el pipeline de transformación es necesario identificar qué variables se tratarán como categóricas y cuáles

```python
categoricas = [
    "sexo",
    "educacion",
    "estado_civil"
]

numericas = [
    col for col in X.columns
    if col not in categoricas
]

print("Variables categóricas:")
print(categoricas)

print("\nVariables numéricas:")
print(numericas)
```

4.4 Construcción del pipeline

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
```

variables categóricas

```python
categoricas = [
    "sexo",
    "educacion",
    "estado_civil"
]
```

Pipeline numérico

```python
num_pipe = Pipeline([
("imputer", SimpleImputer(strategy="median")),
("scaler", StandardScaler())
])
```

Pipeline categórico

```python
cat_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])
```

ColumnTransformer

```python
transformaciones = ColumnTransformer([
    ("num", num_pipe, numericas),
    ("cat", cat_pipe, categoricas)
])
```

```python
pipeline_preparacion = Pipeline([
    ("limpieza", preprocess_transform),
    ("transformacion", transformaciones)
])
```

```python
X_transformado = pipeline_preparacion.fit_transform(X)
```
